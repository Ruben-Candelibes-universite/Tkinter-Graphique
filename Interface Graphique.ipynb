# **Interface Graphique Utilisateur**

<u> Exemple 1 : </u>

```python
from tkinter import *
from tkinter import ttk

# création d'une fenêtre
fenetre=Tk()

# création d'un cadre dans la neêtre et d'une grille
frm=ttk.Frame(fenetre)
frm.grid()

#création d'un texte en vue d'un placement dans le cadre
objet=ttk.Label(frm,text="Essayons d'ourir une fenêtre pour y afficher un message.")

#affichage du texte dans le cadre, selon les coordonnées précisées dans la grille
objet.grid(column=0,row=0)

fenetre.mainloop()
```

<u> Exemple 2 : </u>

```python
# On importe le module Tkinter (TK interface)
from tkinter import *

def affich():
    text3 = Label(
        fenetre,
        text=saisie.get(),
        font=("Arial", 30, "bold")
    )
    text3.pack()

# Création d'une fenêtre (racine de l'interface)
fenetre = Tk()
fenetre.title('Mon application')
fenetre.geometry('500x300+600+200')
f_text = ("Arial", 10)
f_btn = ("Arial", 15, "bold")



# Création d'un texte
text1 = Label(
    fenetre,
    text="Essayons d'ouvrir une fenêtre pour y afficher un message.",
    font=f_text
            )

text2 = Label(
    fenetre,
    text="Ajoutons une deuxième ligne à notre message.",
    font=f_text
            )

saisie = Entry(fenetre,
               width=30,
               font=f_text
              )

btn = Button(fenetre,
             text="Cliquez pour valider",
             font=f_btn,
             command=affich
            )

# Affichage du texte dans la fenêtre - placement automatique
text1.pack()
text2.pack()
saisie.pack()
btn.pack(pady=20)
recup = saisie.get()


# Démarrage de la boucle Tkinter ( s'interrompt lorsqu'on ferme la fenêtre)
fenetre.mainloop()
```

<u> Exemple 3 : </u>

```python
from tkinter import *
from tkinter import ttk

fen = Tk()

frm = ttk.Frame(fen)
frm.grid()

mon_texte1 = ttk.Label(frm, text="Première valeur")
mon_texte1.grid(column=0, row=0, padx=20, pady=20)
mon_texte2 = ttk.Label(frm, text="Deuxième valeur")
mon_texte2.grid(column=0, row=1, padx=20, pady=20)

valeur1 = ttk.Entry(frm)
valeur1.grid(column=1, row=0, padx=20)
valeur2 = ttk.Entry(frm)
valeur2.grid(column=1, row=1, padx=20)

def calcul() :
    somme = int(valeur1.get()) + int(valeur2.get())

    resultat = ttk.Label(frm, text=f"La somme des 2 valeurs est : {somme}")
    resultat.grid(column=2, row=1, padx=10)
    
btn = ttk.Button(frm, text= "Valider les saisies", command = calcul)
btn.grid(column=2, row=0, padx=50)


fen.mainloop()
```
